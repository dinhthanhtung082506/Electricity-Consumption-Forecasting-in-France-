### File french_holidays.csv

```python
import pandas as pd
from pathlib import Path

file_path = Path(r"C:\Users\ACER\Downloads\Electricity Consumption Forecasting in France 2012-2026\Electricity-Consumption-Forecasting-in-France-2012-2026-main\Data\Raw\french_holidays.csv")

df_3 = pd.read_csv(file_path)
print(df_3.head())
```

```python
df_3.columns
```

```python
if 'holiday_name' in df_3.columns:
    df_3 = df_3.drop(columns=['holiday_name'])
```

```python
df_3 = df_3.rename(columns={
    'date': 'Ngày',
    'is_holiday': 'Ngày_lễ'
})
```

```python
df_3.head()
```

```python
df_3['Ngày'] = pd.to_datetime(df_3['Ngày'])

start_date = df_3['Ngày'].min().strftime('%Y-%m-%d')
end_date = df_3['Ngày'].max().strftime('%Y-%m-%d')
```

```python
print(f"Ngày bắt đầu: {start_date}")
print(f"Ngày kết thúc: {end_date}")
```

```python
full_calendar = pd.DataFrame({
    'Ngày': pd.date_range(start=start_date, end=end_date, freq='D')
})
```

```python
full_calendar['Ngày'] = pd.to_datetime(full_calendar['Ngày'])
df_3['Ngày'] = pd.to_datetime(df_3['Ngày'])
```

```python
df_3 = pd.merge(full_calendar, df_3, on='Ngày', how='left')
```

```python
df_3['Ngày_lễ'] = df_3['Ngày_lễ'].fillna(0).astype(int)
```

```python
df_3.tail()
```

### Gộp File

```python
for df in [df_1, df_2, df_3]:
    df["Ngày"] = pd.to_datetime(df["Ngày"])
    df.drop(df[
        (df["Ngày"] < "2012-01-01") |
        (df["Ngày"] > "2026-05-25")
    ].index, inplace=True)

merged = df_1.merge(df_2, on="Ngày", how="inner")
merged = merged.merge(df_3, on="Ngày", how="inner")

merged = merged.sort_values("Ngày").reset_index(drop=True)

print(merged)
print("Số dòng:", len(merged))
```

```python
from pathlib import Path

output_path = Path(r"C:\Users\ACER\Downloads\Electricity Consumption Forecasting in France 2012-2026\Electricity-Consumption-Forecasting-in-France-2012-2026-main\Data\Raw")
output_path.mkdir(parents=True, exist_ok=True)

merged.to_csv(output_path / "raw.csv", index=False, encoding="utf-8-sig")
```
