Chuẩn hóa kiểu dữ liệu

```python
import pandas as pd

# 1. Chuyển đổi cột Ngày sang dạng datetime
df['Ngày'] = pd.to_datetime(df['Ngày'])

# 2. Chuyển đổi cột Ngày_lễ sang dạng category (hoặc giữ nguyên int nếu muốn)
df['Ngày_lễ'] = df['Ngày_lễ'].astype('category')

# 3. Danh sách các cột kiểu float khớp chính xác với bảng của bạn
float_cols = [
    'Tổng tiêu thụ điện (MWh)', 'Nhiệt_độ', 
    'Độ_ẩm', 'Độ_che_phủ_mây', 
    'Bức_xạ_sóng_ngắn', 'Tốc_độ_gió', 
    'Tổng_lượng_mưa'
]
df[float_cols] = df[float_cols].astype('float32')

# 4. Chuyển cột Số_giờ_hợp_lệ sang int32
df['Số_giờ_hợp_lệ'] = df['Số_giờ_hợp_lệ'].astype('int32')

# Kiểm tra lại thông tin sau khi chuyển đổi
print(df.info())
```

Kiểm tra schema

```python
# Danh sách các cột 
num_cols = [
    'Tổng tiêu thụ điện (MWh)', 'Nhiệt_độ', 
    'Độ_ẩm', 'Độ_che_phủ_mây', 
    'Bức_xạ_sóng_ngắn', 'Tốc_độ_gió', 
    'Tổng_lượng_mưa'
]

# Kiểm tra xem có cột số nào chứa dấu phẩy ',' (đặc trưng của định dạng châu Âu) hay không
has_european_comma = any(df[col].astype(str).str.contains(',').any() for col in num_cols)

if has_european_comma:
    print("Phát hiện lỗi: Có dấu phẩy phân cách thập phân kiểu châu Âu trong cột số!")
else:
    print("Thành công: Không có dấu phẩy hay ký tự đặc biệt nào, các cột số hoàn toàn chuẩn sạch!")
```

Đổi tên và chuẩn hóa tất cả các cột kèm đơn vị đo lường chuẩn

```python

df.rename(columns={
    'Ngày': 'Ngày',
    'Tổng tiêu thụ điện (MWh)': 'Tiêu thụ điện (MWh)',
    'Nhiệt_độ': 'Nhiệt độ(°C)',
    'Độ_ẩm': 'Độ ẩm(%)',
    'Độ_che_phủ_mây': 'Độ che phủ mây (%)',
    'Bức_xạ_sóng_ngắn': 'Bức xạ sóng ngắn (W/m²)',
    'Tốc_độ_gió': 'Tốc độ gió (km/h)',
    'Tổng_lượng_mưa': 'Tổng lượng mưa (mm)',
    'Số_giờ_hợp_lệ': 'Số giờ hợp lệ (h)',
    'Ngày_lễ': 'Ngày lễ (0/1)'
}, inplace=True)

```

```python
df = df.drop(columns=['Số giờ hợp lệ (h)'])
```

Chuẩn hóa định dạng số và làm tròn số thập phân

```python

# 1. Chuyển đổi cột Ngày sang định dạng datetime chuẩn
df['Ngày'] = pd.to_datetime(df['Ngày'])

# 2. Danh sách các cột chứa số thập phân cần làm tròn (lấy 2 chữ số thập phân)
float_cols = [
    'Tiêu thụ điện (MWh)', 
    'Nhiệt độ(°C)', 
    'Độ ẩm(%)', 
    'Độ che phủ mây (%)', 
    'Bức xạ sóng ngắn (W/m²)', 
    'Tốc độ gió (km/h)', 
]

# Làm tròn các cột float đến 2 chữ số thập phân
df[float_cols] = df[float_cols].round(2)

# 3. Chuẩn hóa các cột số nguyên/đếm (Số giờ hợp lệ, Ngày lễ)
df['Ngày lễ (0/1)'] = df['Ngày lễ (0/1)'].astype(int)
```

Định nghĩa các điều kiện ràng buộc range

```python
df.min_mwh = df['Tiêu thụ điện (MWh)'].min()
df.max_mwh = df['Tiêu thụ điện (MWh)'].max()
print(f"Giá trị nhỏ nhất của Tiêu thụ điện (MWh): {df.min_mwh}")
print(f"Giá trị lớn nhất của Tiêu thụ điện (MWh): {df.max_mwh}")
```

```python
# Tìm dòng có giá trị min và max
min_row = df.loc[df['Tiêu thụ điện (MWh)'].idxmin()]
max_row = df.loc[df['Tiêu thụ điện (MWh)'].idxmax()]

# In kết quả kèm theo ngày tương ứng
print(f"Ngày tiêu thụ thấp nhất: {min_row['Ngày']} ({min_row['Tiêu thụ điện (MWh)']:.2f} MWh)")
print(f"Ngày tiêu thụ cao nhất: {max_row['Ngày']} ({max_row['Tiêu thụ điện (MWh)']:.2f} MWh)")
```

```python
limits = {
    'Tiêu thụ điện (MW)': (20000, 110000),  # Mở rộng phù hợp quy mô toàn quốc gia Pháp
    'Nhiệt độ(°C)': (-30, 45),
    'Độ ẩm(%)': (0, 100),
    'Độ che phủ mây (%)': (0, 100),
    'Bức xạ sóng ngắn (W/m²)': (0, 1400),
    'Tốc độ gió (km/h)': (0, 200),
    'Tổng lượng mưa (mm)': (0, 300),
    'Ngày lễ (0/1)': (0, 1)
}

for col, (mn, mx) in limits.items():
    if col in df.columns:
        print(f"{col}: {len(df[(df[col] < mn) | (df[col] > mx)])} vi phạm")
```

Kiểm tra ngoại lai

```python
df.columns
```

```python
total_rows = len(df)
for col in ['Tiêu thụ điện (MWh)', 'Nhiệt độ(°C)', 'Độ ẩm(%)', 'Độ che phủ mây (%)', 'Bức xạ sóng ngắn (W/m²)', 'Tốc độ gió (km/h)', 'Tổng lượng mưa (mm)']:
    Q1, Q3 = df[col].quantile(0.25), df[col].quantile(0.75)
    iqr = Q3 - Q1
    outliers_count = len(df[(df[col] < Q1 - 1.5 * iqr) | (df[col] > Q3 + 1.5 * iqr)])
    pct = (outliers_count / total_rows) * 100
    print(f"{col}: {outliers_count} ngoại lai ({pct:.2f}%)")
```

Lưu ý: Dùng các mô hình học máy ít bị ảnh hưởng bởi outlier (như các mô hình dựa trên cây quyết định như Random Forest, XGBoost, LightGBM) hoặc sử dụng các hàm mất mát (loss functions) ít nhạy cảm với ngoại lai khi huấn luyện.
